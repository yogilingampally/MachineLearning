```python
import numpy as np
import statistics
```

```python
class KNN_classifier:
  #initiating parameters
  def __init__(self, distance_matric):
    self.distance_matrix = distance_matric

  def get_distance_matric(self, training_data_point, test_data_point):
    if self.distance_matrix == 'euclidean':
      dist = 0
      # Exclude the last element if it's the label
      for i in range(len(training_data_point) - 1):
        dist += (training_data_point[i] - test_data_point[i])**2
      euclidean_dist = np.sqrt(dist)
      return euclidean_dist
    elif self.distance_matrix == 'manhattan':
      dist = 0
      # Exclude the last element if it's the label
      for i in range(len(training_data_point) - 1):
        dist += abs(training_data_point[i] - test_data_point[i])
      manhattan_dist = dist
      return manhattan_dist
    else:
      raise ValueError("Invalid distance matrix specified. Choose 'euclidean' or 'manhattan'.")

  def nearest_neighbors(self, X_train, test_data, k):
    distance_list = []
    for training_data_idx in range(len(X_train)):
      distance = self.get_distance_matric(X_train[training_data_idx], test_data)
      # Store (distance, index of training data point)
      distance_list.append((distance, training_data_idx))
    distance_list.sort(key=lambda x: x[0]) # Sort by distance

    neighbors_indices = []
    for j in range(k):
      neighbors_indices.append(distance_list[j][1]) # Append the index of the k nearest neighbors
    return neighbors_indices # Returns indices of k nearest neighbors

  def predict(self, X_train, test_data, k):
    # Get the indices of the k nearest neighbors
    neighbor_indices = self.nearest_neighbors(X_train, test_data, k)

    labels = []
    for idx in neighbor_indices:
      # Assuming the last element of X_train[idx] is the label
      labels.append(X_train[idx][-1])

    if not labels:
      # Handle case where no neighbors were found or labels list is empty
      return None

    predicted_class = statistics.mode(labels)
    return predicted_class
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data=pd.read_csv('/content/diabetes.csv')
```

```python
X=data.drop(columns='Outcome',axis=1)
Y=data['Outcome']
```

```python
X=X.to_numpy()
Y=Y.to_numpy()
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.2,stratify=Y,random_state=2)
```

```python
print(Y_train)
```

```python
X_train=np.insert(X_train,8,Y_train,axis=1)
```

```python
classifier=KNN_classifier('euclidean')
```

```python
prediction=classifier.predict(X_train,X_test[0],k=5)
```
